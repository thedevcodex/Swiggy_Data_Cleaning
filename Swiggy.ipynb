# Data Cleaning

## Change OrderDate DataType

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'],dayfirst=True,errors='coerce')
df['OrderDate'] = df['OrderDate'].dt.strftime('%d-%m-%Y')
```

## Clean Inconsistent Values on Gender

```python
df['Gender'] = df['Gender'].str.lower()

df['Gender'] = df['Gender'].map({
    'm':'Male',
    'male':'Male',
    'f':'Female',
    'female':'Female'
})
```

```python
df.head(20)
```

```python
df['City'].unique()
```

## Column City

```python
df['City'] = df['City'].str.lower().str.strip()

df['City'] = df['City'].map({
    "mumbai":"Mumbai",
    "bangalore":"Bangalore",
    "bengaluru":"Bangalore",
    'kolkata':"Kolkata",
    'pune':'Pune',
    'hyderabad':'Hyderabad',
    'chennai':'Chennai',
    'delhi':'Delhi'
})
```

```python
df.head(20)
```

## TotalAmount

```python
df['TotalAmount'] = (df['TotalAmount']
    .astype(str)
    .str.replace("Rs.","",regex=False)
    .str.replace("₹","",regex=False)
    .str.strip()
    .astype(float)
)
```

```python
df['TotalAmount'].dtype
```

```python
df.columns
```

## Cuisine

```python
df['Cuisine'].unique()
```

```python
df['Cuisine'] = df['Cuisine'].str.lower()
df['Cuisine'] = df['Cuisine'].map({
    "bakery":"Bakery",
    "chinese":"Chinese",
    "japanese":"Japanese",
    "street food":"Street Food",
    "fast food":"Fast Food",
    "briyani":"Briyani",
    'italian':"Italian",
    'north indian':'North Indian',
    'south indian':"South Indian",
    'salads':'Salads',
    'desserts':"Desserts",
    'mughlai':'Mughlai',
    'mexican':'Mexican',
    'north-indian':"North Indian"

})
```

```python
df['Cuisine'].unique()
```

## UnitPrice

```python
df['UnitPrice'].unique()
```

```python
df['UnitPrice'] = (
    df['UnitPrice']
    .astype(str)
    .str.replace("Rs.","",regex=False)
    .str.replace("₹","",regex=False)
    .str.strip()
    .astype(float)
)
```

## OrderStatus

```python
df['OrderStatus'] = df['OrderStatus'].str.lower()

df['OrderStatus'] = df['OrderStatus'].map({
    "delivered":"Delivered",
    "returned":"Returned",
    "cancelled":"Cancelled"
})
```

```python
df['OrderStatus'].unique()
```

## PaymentMode

```python
df['PaymentMode'] = df['PaymentMode'].str.lower()
```

```python
df[df['PaymentMode']=="-"] = np.nan
```

```python
df['PaymentMode'].unique()
```

```python
df['PaymentMode'] = df['PaymentMode'].str.lower()
df['PaymentMode'] = df['PaymentMode'].map({
    "wallet":"Wallet",
    "debitcard":"DebitCard",
    "credit card":"CreditCard",
    'upi':'UPI',
    'cash on delivery':'Cash On Delivery',
    'creditcard':'CreditCard',
    'debit card':'DebitCard',
    'cashondelivery':"Cash On Delivery"
})
```

## Clean RestaurantName

```python
df['RestaurantName']=df['RestaurantName'].str.strip()
```

## Handle Missing Values

### Numerical columns

```python
num_cols = df.select_dtypes(include='number').columns

for i in num_cols:
  if -0.5<= df[i].skew() <=0.5:
    df[i] = df[i].fillna(df[i].mean())
  else:
    df[i] = df[i].fillna(df[i].median())
```

### Categorical columns

```python
cat_cols = df.select_dtypes(include='object').columns

for i in cat_cols:
  df[i] = df[i].fillna(df[i].mode()[0])
```

```python
df.isnull().sum()
```

## Duplicate Values

```python
df = df.drop_duplicates()
```
